```python
## Analys med One-Hot Encoded Features

Nu när vi har skapat one-hot encoded features kan vi göra djupare analyser.
```

```python
# Korrelationsanalys: Hur korrelerar one-hot features med total_kop?
# Välj bara numeriska kolumner för korrelationsberäkning

numeric_cols = df_feat.select_dtypes(include=[np.number]).columns
correlation_with_total_kop = df_feat[numeric_cols].corr()["total_kop"].sort_values(ascending=False)
print("Korrelation med total_kop:\n")
print(correlation_with_total_kop)
```

```python
# Filtrera ut bara stad och medlemsnivå-features för tydligare bild
stad_cols = [col for col in df_feat.columns if col.startswith("stad_")]
medlems_cols = [col for col in df_feat.columns if col.startswith("medlemsniva_")]

print("=" * 50)
print("STAD-FEATURES KORRELATION MED TOTAL_KÖP:")
print("=" * 50)
stad_corr = df_feat[stad_cols + ["total_kop"]].corr()["total_kop"].drop("total_kop").sort_values(ascending=False)
print(stad_corr)

print("\n" + "=" * 50)
print("MEDLEMSNIVÅ-FEATURES KORRELATION MED TOTAL_KÖP:")
print("=" * 50)
medlems_corr = df_feat[medlems_cols + ["total_kop"]].corr()["total_kop"].drop("total_kop").sort_values(ascending=False)
print(medlems_corr)
```

```python
# Genomsnittligt köpbelopp per stad (med one-hot encoding)
print("GENOMSNITTLIGT TOTAL_KÖP PER STAD:")
print("-" * 40)

for col in stad_cols:
    stad_namn = col.replace("stad_", "").title()
    avg_kop = df_feat[df_feat[col] == 1]["total_kop"].mean()
    antal = df_feat[df_feat[col] == 1].shape[0]
    print(f"{stad_namn:15} | Snitt: {avg_kop:,.0f} kr | Antal kunder: {antal}")
```

```python
# Genomsnittligt köpbelopp per medlemsnivå (med one-hot encoding)
print("GENOMSNITTLIGT TOTAL_KÖP PER MEDLEMSNIVÅ:")
print("-" * 40)

for col in medlems_cols:
    niva_namn = col.replace("medlemsniva_", "").title()
    avg_kop = df_feat[df_feat[col] == 1]["total_kop"].mean()
    median_kop = df_feat[df_feat[col] == 1]["total_kop"].median()
    antal = df_feat[df_feat[col] == 1].shape[0]
    print(f"{niva_namn:10} | Snitt: {avg_kop:,.0f} kr | Median: {median_kop:,.0f} kr | Antal: {antal}")
```

```python
# Skapa en sammanfattningstabell med aggregerade statistik per one-hot feature
def analyze_one_hot_features(df, feature_cols, target_col):
    """Analyserar one-hot encoded features mot en målvariabel"""
    results = []
    
    for col in feature_cols:
        grupp = df[df[col] == 1][target_col]
        results.append({
            "feature": col,
            "antal": grupp.count(),
            "medel": grupp.mean(),
            "median": grupp.median(),
            "std": grupp.std(),
            "min": grupp.min(),
            "max": grupp.max()
        })
    
    return pd.DataFrame(results).sort_values("medel", ascending=False)

# Analysera stad-features
print("DETALJERAD ANALYS - STAD FEATURES:")
stad_analysis = analyze_one_hot_features(df_feat, stad_cols, "total_kop")
stad_analysis
```

```python
# Analysera medlemsnivå-features
print("DETALJERAD ANALYS - MEDLEMSNIVÅ FEATURES:")
medlems_analysis = analyze_one_hot_features(df_feat, medlems_cols, "total_kop")
medlems_analysis
```

```python
# Korsanalys: Stad x Medlemsnivå - Antal kunder i varje kombination
# Använder originaldata för korsanalys
print("KORSANALYS: ANTAL KUNDER PER STAD OCH MEDLEMSNIVÅ")
print("=" * 60)

cross_tab = pd.crosstab(df["stad"], df["medlemsniva"], margins=True, margins_name="Totalt")
cross_tab
```

```python
# Pivot-tabell: Genomsnittligt köpbelopp per Stad x Medlemsnivå
print("GENOMSNITTLIGT KÖPBELOPP PER STAD OCH MEDLEMSNIVÅ")
print("=" * 60)

pivot_avg = df.pivot_table(
    values="total_kop", 
    index="stad", 
    columns="medlemsniva", 
    aggfunc="mean"
).round(0)

pivot_avg
```

```python
# Identifiera bästa kombinationer baserat på one-hot features
print("TOP KOMBINATIONER - STAD + MEDLEMSNIVÅ MED HÖGST GENOMSNITTLIGT KÖP")
print("=" * 70)

# Skapa kombinationsanalys
combo_analysis = df.groupby(["stad", "medlemsniva"]).agg({
    "total_kop": ["mean", "sum", "count"]
}).round(0)

combo_analysis.columns = ["Snitt_kop", "Total_kop", "Antal_kunder"]
combo_analysis = combo_analysis.sort_values("Snitt_kop", ascending=False)
combo_analysis
```

```python
# Feature Importance Analys (baserat på korrelationsstyrka)
# Skapar en "feature importance"-liknande ranking

all_one_hot_cols = stad_cols + medlems_cols
feature_importance = df_feat[all_one_hot_cols + ["total_kop"]].corr()["total_kop"].drop("total_kop")

# Skapa en DataFrame med absolut korrelation för ranking
importance_df = pd.DataFrame({
    "feature": feature_importance.index,
    "korrelation": feature_importance.values,
    "abs_korrelation": np.abs(feature_importance.values)
}).sort_values("abs_korrelation", ascending=False)

print("FEATURE IMPORTANCE RANKING (baserat på korrelation med total_kop)")
print("=" * 70)
print("\nPositiv korrelation = köper MER")
print("Negativ korrelation = köper MINDRE")
print("-" * 70)
importance_df
```
