```python
def clean_date(df):

    df["order_datum"] = pd.to_datetime(df["order_datum"], dayfirst=True, errors="coerce", format="mixed")

    return df




```

```python
def add_order_total(df):
    
    df["order_total"] = df["pris_st"] * df["antal"]

    return df
```

```python
def add_time_features(df):

    df["order_month"] = df["order_datum"].dt.month
    df["order_day"] = df["order_datum"].dt.day_name()

    return df

```

```python
def transform_orders(df):
    """
    Komplett transform-pipeline för orderdata.
    """
    print("Startar transform...")

    df_clean = df.copy()

    df_clean = clean_categories(df_clean)
    df_clean = clean_channel(df_clean)
    df_clean = clean_status(df_clean)
    df_clean = clean_order_price(df_clean)
    df_clean = clean_date(df_clean)

    df_clean = add_order_total(df_clean)    
    df_clean = add_time_features(df_clean)

    print("Transform klar!")

    return df_clean
```
